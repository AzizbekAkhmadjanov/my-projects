```python
import requests

username = "Azizbek4254"
url = f"https://api.chess.com/pub/player/{username}/games/archives"
response = requests.get(url, headers={"User-Agent": "Azizbek4254 - personal data pull"})
archives = response.json()["archives"]
print(archives)
```

```python
all_games = []
for archive_url in archives:
    resp = requests.get(archive_url, headers={"User-Agent": "Azizbek4254 - personal data pull"})
    all_games.extend(resp.json()["games"])

print(f"Total games: {len(all_games)}")
```

```python
import json
print(json.dumps(all_games[0], indent=2))
```

```python
import re

PGN_TAG_RE = re.compile(r'\[(\w+)\s+"(.*?)"\]')
```

```python
def parse_pgn_tags(pgn: str) -> dict:
    return dict(PGN_TAG_RE.findall(pgn or ""))
```

```python
tags = parse_pgn_tags(all_games[0]["pgn"])
print(tags)
```

```python
game = all_games[0]

easy_fields = {
    "time_class": game.get("time_class"),
    "rated": game.get("rated"),
    "rules": game.get("rules"),
    "end_time": game.get("end_time"),
    "white_username": game.get("white", {}).get("username"),
    "white_rating": game.get("white", {}).get("rating"),
    "white_result": game.get("white", {}).get("result"),
    "black_username": game.get("black", {}).get("username"),
    "black_rating": game.get("black", {}).get("rating"),
    "black_result": game.get("black", {}).get("result"),
}

print(easy_fields)
```

```python
row = {**easy_fields, **parse_pgn_tags(game["pgn"])}
print(row)
```

```python
rows = []

for game in all_games:
    easy_fields = {
        "time_class": game.get("time_class"),
        "rated": game.get("rated"),
        "rules": game.get("rules"),
        "end_time": game.get("end_time"),
        "white_username": game.get("white", {}).get("username"),
        "white_rating": game.get("white", {}).get("rating"),
        "white_result": game.get("white", {}).get("result"),
        "black_username": game.get("black", {}).get("username"),
        "black_rating": game.get("black", {}).get("rating"),
        "black_result": game.get("black", {}).get("result"),
    }
    row = {**easy_fields, **parse_pgn_tags(game["pgn"])}
    rows.append(row)

print(len(rows))
print(rows[0])
```

```python
import pandas as pd

df = pd.DataFrame(rows)
print(df.shape)
df.head()
```

```python
print(df.columns.tolist())
```

```python
df.to_csv("all_games_full.csv", index=False)
```

```python
df = df.drop(columns=[
    "rated",
    "rules",
    "Event",
    "Site",
    "Round",
    "White",
    "Black",
    "CurrentPosition",
    "Timezone",
    "ECO",
    "ECOUrl",
    "UTCDate",
    "UTCTime",
    "Link",
])

print(df.shape)
df.head()
```

```python
df.to_csv("chess_games_clean.csv", index=False)
```

```python
df["start_datetime"] = pd.to_datetime(df["Date"] + " " + df["StartTime"], format="%Y.%m.%d %H:%M:%S")
df["end_datetime"] = pd.to_datetime(df["EndDate"] + " " + df["EndTime"], format="%Y.%m.%d %H:%M:%S")

df[["start_datetime", "end_datetime"]].head()
```

```python
df["duration"] = df["end_datetime"] - df["start_datetime"]

df["duration"].head()
```

```python
df["duration_seconds"] = df["duration"].dt.total_seconds()
df["duration_minutes"] = df["duration_seconds"] / 60

df[["duration", "duration_seconds", "duration_minutes"]].head()
```

```python
df["duration_minutes"].mean()   # average game length across all 2843 games
df["duration_minutes"].max()    # your longest game
df.sort_values("duration_minutes", ascending=False).head()  # your top 5 longest games
```
